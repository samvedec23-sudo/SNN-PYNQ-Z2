```python
import os
import pickle
import numpy as np

# Locate the test_batch file inside the cifar-10-python directory structure
possible_paths = [
    "cifar-10-python/test_batch",
    "cifar-10-python/cifar-10-batches-py/test_batch",
    "./test_batch"
]

test_batch_path = None
for path in possible_paths:
    if os.path.exists(path):
        test_batch_path = path
        break

# If not found directly, search recursively
if test_batch_path is None:
    for root, dirs, files in os.walk("cifar-10-python"):
        if "test_batch" in files:
            test_batch_path = os.path.join(root, "test_batch")
            break

if test_batch_path is None:
    raise FileNotFoundError("Could not find 'test_batch'. Please ensure 'cifar-10-python' is fully extracted.")

print(f"Found CIFAR-10 test batch at: {test_batch_path}")

# Load the test batch using pickle
with open(test_batch_path, "rb") as f:
    batch = pickle.load(f, encoding="bytes")

# Reshape raw image data into standard (N, 32, 32, 3) format
x_test = batch[b'data'].reshape(-1, 3, 32, 32).transpose(0, 2, 3, 1)
y_test = np.array(batch[b'labels'])

# Select 50 test samples for your live presentation demo
num_samples = 50
test_images = x_test[:num_samples]
test_labels = y_test[:num_samples]

# Convert test images into 4096-feature sparse binary spike matrices
np.random.seed(42)
flat_gray = np.mean(test_images, axis=-1) / 255.0
test_features = np.zeros((num_samples, 4096), dtype=np.uint32)
for i in range(num_samples):
    resized = np.resize(flat_gray[i], (64, 64)).flatten()
    test_features[i] = (resized > 0.5).astype(np.uint32)

# Save the .npy files into the current working directory
np.save("cifar10_test_images.npy", test_images)
np.save("cifar10_test_features.npy", test_features)
np.save("cifar10_test_labels.npy", test_labels)

print("Successfully generated and verified:")
print(f" - cifar10_test_images.npy   -> shape: {test_images.shape}")
print(f" - cifar10_test_features.npy -> shape: {test_features.shape}")
print(f" - cifar10_test_labels.npy   -> shape: {test_labels.shape}")
```

```python
import os
import tarfile
import pickle
import numpy as np

# 1. Automatically locate and extract the tar.gz archive if not already done
archive_name = "cifar-10-python.tar.gz"
if os.path.exists(archive_name):
    print(f"Found {archive_name}, extracting contents...")
    with tarfile.open(archive_name, "r:gz") as tar:
        tar.extractall()
    print("Extraction complete!")
else:
    print(f"Note: {archive_name} not in current directory, searching existing folders...")

# 2. Locate the test_batch file dynamically
test_batch_path = None
for root, dirs, files in os.walk("."):
    if "test_batch" in files:
        test_batch_path = os.path.join(root, "test_batch")
        break

if test_batch_path is None:
    raise FileNotFoundError("Could not locate 'test_batch' anywhere in the workspace. Please check if cifar-10-python.tar.gz is in this folder.")

print(f"Found CIFAR-10 test batch at: {test_batch_path}")

# 3. Load the test batch using pickle
with open(test_batch_path, "rb") as f:
    batch = pickle.load(f, encoding="bytes")

# 4. Reshape and process into numpy arrays
x_test = batch[b'data'].reshape(-1, 3, 32, 32).transpose(0, 2, 3, 1)
y_test = np.array(batch[b'labels'])

num_samples = 50
test_images = x_test[:num_samples]
test_labels = y_test[:num_samples]

np.random.seed(42)
flat_gray = np.mean(test_images, axis=-1) / 255.0
test_features = np.zeros((num_samples, 4096), dtype=np.uint32)
for i in range(num_samples):
    resized = np.resize(flat_gray[i], (64, 64)).flatten()
    test_features[i] = (resized > 0.5).astype(np.uint32)

# 5. Save the output .npy files
np.save("cifar10_test_images.npy", test_images)
np.save("cifar10_test_features.npy", test_features)
np.save("cifar10_test_labels.npy", test_labels)

print("Success! Generated files:")
print(" - cifar10_test_images.npy")
print(" - cifar10_test_features.npy")
print(" - cifar10_test_labels.npy")
```
